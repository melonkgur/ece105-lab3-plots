```python
import numpy as np
import matplotlib.pyplot as plt

# Generate simulated temperature readings from two sensors.
# Sensor A: mean 25 C, std 3 C, 200 readings.
# Sensor B: mean 27 C, std 4.5 C, 200 readings.
# Also generate 200 timestamps uniformly from 0 to 10 seconds.
# Use np.random.default_rng with a seed = last 4 digits of your Drexel ID.
seed = 3557
rng = np.random.default_rng(seed)

n = 200
t = rng.uniform(0, 10, n)
t.sort()  # make timestamps increasing for time-series plots

sensor_A = rng.normal(loc=25.0, scale=3.0, size=n)
sensor_B = rng.normal(loc=27.0, scale=4.5, size=n)

```

```python
print(sensor_A.shape)
```

```python
# Use sensor_a, sensor_b, timestamps as specified
sensor_a = sensor_A.astype(np.float64)
sensor_b = sensor_B.astype(np.float64)
timestamps = t.astype(np.float64)

plt.figure(figsize=(8,5))
plt.scatter(timestamps, sensor_a, c='C0', s=25, alpha=0.7, label='Sensor A')
plt.scatter(timestamps, sensor_b, c='C1', s=35, alpha=0.7, label='Sensor B', marker='x')
plt.xlabel('Time (s)')
plt.ylabel('Temperature (°C)')
plt.title('Scatter of sensor readings vs time')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
